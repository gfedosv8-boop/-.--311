```python
def is_year_leap(year):
  if year % 400 == 0:
    return True
  if year % 100 == 0:
    return False
  if year % 4 == 0:
    return True
  return False
is_year_leap(2025)
```

```python
def is_prime(n):
  if n % 2 == 0:
    return False
  if n == 2:
    return True
  if n < 2 == 0:
    return False
  return True
is_prime(13)

```

```python
numbers = []
for i in range(5):
  numbers.append(int(input("Введите число:")))
if len(set(numbers)) == 5:
  print("Все числа уникальные")
else:
  print("Есть одинаковые числа")
even = 0
negative = 0
in_range = 0
for num in numbers:
  if num % 2 == 0:
    even = even + 1
  if num < 0:
    negative = negative + 1
  if -256 <= num <= 1024:
    in_range = in_range + 1
print("Четных:" , even)
print("Отрицательных:" , negative)
print("В промежутке [-256 ; 1024]:" , in_range)


```

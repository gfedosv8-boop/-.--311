```python
a = float(input("Сторона a:"))
b = float(input("Сторона b:"))
c = float(input("Сторона c:"))
if a + b == c and a + c == b and b + c == a:
  perimetr = a + b + c
  p = perimetr / 2
  area = (p * (p - a) * (p - b ) * (p - c)) ** 0.5
  print("Периметр:" , perimetr)
  print("Площадь:" , area)
else:
  print("Треугольника с такими сторонами не существует")

```

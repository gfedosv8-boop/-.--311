```python
import matplotlib.pyplot as plt
celsius = [[-67.0], [-34.0], [0], [34.0], [54.0], [67.0], [100.0]]
fahrenheit = [[-88.6], [-29.2], [32.0], [93.2], [129.2],  [152.6], [212.0]]
plt.figure(figsize=(15,8), dpi=50)
plt.scatter(celsius , fahrenheit , label = 'входные значения' , color = 'green' , marker = '$f$')
plt.xlabel('celsius')
plt.ylabel('fahrenheit')
plt.legend()
plt.grid(True)
plt.show()
for c , f in zip(celsius , fahrenheit):
  print(f'цельсия {c}= фаренгейт {f}')
  from sklearn.linear_model import LinearRegression
  lr = LinearRegression()
  lr.fit(celsius , fahrenheit)
  lr.predict ([[256] , [123]])
  lr.coef_
  lr.intercept_
  celsius_test= [[-50] , [10] , [30] , [20] , [10] , [70] , [87]]
  fahrenheit_test= lr.predict(celsius_test)
  fahrenheit_test
  for c , f in zip(celsius_test , fahrenheit_test):
    print(f'цельсия {c} фаренгейта {f}')
import numpy as np
x_range = np.arange(-70 , 120)
y_range = x_range*1.8+32
plt.figure(figsize=(15,8), dpi=280)
plt.plot(x_range , y_range , label ='уравнение' , linewidth =1)
plt.scatter(celsius , fahrenheit , label='входные данные' , color='green')
plt.scatter(celsius_test , fahrenheit_test , label='предсказанное значение' , color='blue')
plt.xlabel('Цельсия')
plt.ylabel('Фаренгейта')
plt.legend()
plt.grid(True)
plt.show()

```

```python
fahrenheit_k = [[-40.0], [0.0], [32.0], [50.0], [86.0], [122.0], [212.0]]
kelvin = [[round((f[0] - 32) * 5 / 9 + 273.15, 2)] for f in fahrenheit_k]
plt.figure(figsize=(10, 6))
plt.scatter(fahrenheit_k, kelvin, label='входные значения', color='green', marker='$F$')
plt.xlabel('fahrenheit')
plt.ylabel('kelvin')
plt.legend()
plt.grid(True)
plt.show()
for f, k in zip(fahrenheit_k, kelvin):
    print(f'фаренгейт {f} = кельвин {k}')
lr_k = LinearRegression()
lr_k.fit(fahrenheit_k, kelvin)
print('коэффициент:', lr_k.coef_, '(ожидаем 5/9 =', round(5 / 9, 4), ')')
print('свободный член:', lr_k.intercept_)
print(lr_k.predict([[256], [123]]))
fahrenheit_k_test = [[-50], [10], [30], [20], [10], [70], [87]]
kelvin_test = lr_k.predict(fahrenheit_k_test)
for f, k in zip(fahrenheit_k_test, kelvin_test):
    print(f'фаренгейт {f} = кельвин {k}')
x_f = np.arange(-60, 230)
y_k = (x_f - 32) * 5 / 9 + 273.15
plt.figure(figsize=(12, 7), dpi=100)
plt.plot(x_f, y_k, label='уравнение', linewidth=1)
plt.scatter(fahrenheit_k, kelvin, label='выходные значения', color='green')
plt.scatter(fahrenheit_k_test, kelvin_test, label='предсказанное значение', color='blue')
plt.xlabel('Фаренгейт')
plt.ylabel('Кельвин')
plt.legend()
plt.grid(True)
plt.show()
```
